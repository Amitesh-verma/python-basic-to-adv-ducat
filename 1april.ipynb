# Slicing Assignment


```python
# String Slicing
# Ek string "PythonProgramming" di hai:
# • First 6 characters nikaalo
# • Last 5 characters nikaalo

str = "PythonProgramming"
str[0:6]

str[-6:-1]
```

```python
# String "HelloWorld" me se:
# • "World" slice karo
# • Har second character print karo

str = "HelloWorld"
str[5::1]

str[::2]
```

```python
# String ko reverse karo using slicing

str = "HelloWorld"
str[::-1]
```

```python
# "DataScience" me se:
# • Odd index characters nikaalo
# • Even index characters nikaalo

str = "DataScience"

str[0::2] # even index

str[1::2]  # odd index
```

```python
# Ek string input lo aur:
# • Uska middle character (ya 2 middle characters agar length even ho) print karo
str = input()

len(str)

if len(str) % 2:
    print('odd')
else:
    print('even')    

```

```python
# "ABCDEFGHIJ" ko 3-3 ke chunks me slice karo

str =  "ABCDEFGHIJ"

for i in range(0,len(str),3):
    print(str[i:i+3])
```

```python
# String me se first aur last character remove karke new string banao

str  = input()
str[1::]
str[1:-1]
```
