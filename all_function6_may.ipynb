List Comprehension

```python
nums = [1,2,3,4]
result = [] #empty list
# g='' #empty string
for x in nums:  #iteration
    result.append(x*2)

print(result)
```

```python
nums = [1,2,3,4]
result = [x*2 for x in nums]
print(result)
```

```python
nums = [1,2,3,4,5]

result = []

for x in nums:
    if x % 2 == 0:
        result.append(f"even:-{x}")
    else:
        result.append(f"odd:-{x}")

print(result)
```

```python
nums = [1,2,3,4,5]

result = ["even" if x % 2 == 0 else "odd" for x in nums]

print(result)
```

Dictionary Comprehension

```python
nums = [1,2,3,4]

result = {}
# result={
#     1:"odd",
#     2:"even",
#     3:"odd",
#     4:"even"  
# }

for x in nums:
    if x % 2 == 0:
        result[x] = "even"
    else:
        result[x] = "odd"

print(result)
```

```python
nums = [1,2,3,4]

result = {x: ("even" if x%2==0 else "odd") for x in nums}
          # key:value
             # 1:"odd",
             # 2:"even"

print(result)
```

map function

```python
nums = [1, 2, 3]
# map for loop km kardeta hai
result = list(map(lambda x: x + 1, nums))

print(result)
```

filter function 

```python
nums = [1, 2, 3, 4, 5, 6]

result = list(filter(lambda x: x % 2 == 0, nums))

print(result)
```

reduce function

```python
# saare elements ko combine karke ek single value deta hai
from functools import reduce

nums = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, nums)
                       # 1,2:  1+2=3
                       # 3,3: 3+3=6
                       # 6,4: 6+4=10
print(result)
```
