```python
#syntax  for loop

# for iterator in sequence:
#       opeartion\states
i=2
for i in range(1,10):
    print(i,'airplane')
```

```python
# Print numbers from 1 to 10 using a for loop

for i in range(1,11):
    print(i)
```

```python
# Print even numbers between 1 and 50.

for i in range(1,51):
    if i%2==0:
        print(i)
    else:
        pass    
```

```python
# Print odd numbers from 1 to 100.
for i in range(1,101):
    if i%2!=0:
        print(i)
    else:
        pass 
```

```python
# Find the sum of first n natural numbers.

n = int(input())
total = 0
for i in range(1,n+1):
    total = total+i
print('sum is',total)
```

```python
# Print multiplication table of a given number.

num = int(input('Enter the num : '))
for i in range(1,11):
    print(num, 'X', i ,'=' , num * i)
```

```python
# *
# **
# ***
# ****
# *****

for i in range(1, 6):
    print( "*" * i )
```

```python
# Find factorial of a number using a for loop.
n = int(input('Enter the number : '))
fact = 1
for i in range(1,n+1):
    fact = fact * i

print('fact is : ',fact)    
```
