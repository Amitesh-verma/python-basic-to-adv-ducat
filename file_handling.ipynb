```python
# Reading Files
#  Read entire file

with open('notes.txt') as f:
    data = f.read()
    print(data)
```

```python
# Read line by line
with open('notes.txt') as f:
    line  = f.readline()
    print(line)
```

```python
# Read all lines
with open('notes.txt','r') as f:
    line1  = f.readline()
    line2 = f.readline()
    line3 = f.readline()
    line4 = f.readline()
    line5 = f.readline()


    print('line 1',line1)
    print('line 2',line2)
    print('line 3',line3)
    print('line 4',line4)
    print('line 5',line5)

```
