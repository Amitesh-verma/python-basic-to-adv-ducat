# Nested for


```python
adj = ['red','tasty','good']
fruits = ['apple','banana','cherry','mango']
for i in adj:
    for j in fruits:
        print(i,j)
```

```python

num  = 4
for i in range(num):
    for j in range(num):
        print('*',end  = " ")
    print()
```

```python
num  = 4+1
for i in range(num):
    for j in range(i):
        print('*',end  = " ")
    print()
```
