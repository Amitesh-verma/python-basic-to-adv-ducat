```python
# Write a function that takes a string and returns the count of vowels and consonants separately.

def countvowcons(userinput):
    vowels = "aieouAIEOU"

    countvowels = 0
    countconsonants = 0

    for eachchar in userinput:
        if(eachchar.isalpha()):
            if(eachchar in vowels):
                countvowels +=1
            else:
                countconsonants +=1
                
    return countvowels, countconsonants

vowels,consonants = countvowcons("Amitesh")

print(vowels,consonants)

```
