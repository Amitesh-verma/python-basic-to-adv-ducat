```python
# Simple Calculator
#Take two numbers and an operator (+, -, *, /) and perform the operation.

num1 = int(input())
num2 = int(input())
operator = input('which operation you wanted to perform(+.-.*./)')

if operator == '+':
    print(num1+num2)
elif operator ==  '-':
    print(num1-num2)
elif operator == '*':
    print(num1*num2)   
else:
    print(num1/num2)               
```

```python
# Check Leap Year
year = int(input('enter year for checking leap year'))
if(year%4==0 and year%100!=0) or year%400==0:
    print('year',year,'is leap year')
else:
    print('NOT',year,'is leap year')    
```

```python
# Grade System
# Marks input:
# ≥90 → A
# 75–89 → B
# 50–74 → C
# # <50 → Fail

marks = int(input())
if marks >=90:
  print('Grade A')
elif marks >= 75 and marks <= 89:
  print('Grade B')
elif marks >= 50 and marks <= 74:
  print('Grade C') 
else:
  print('Fail')     
```

```python
# ATM Withdrawal
# Input amount:
# Must be multiple of 100
# Balance should be sufficient
# Otherwise show error
 
amount = int(input())
balance = 1000

if amount % 100 == 0 and amount <= balance:
    balance -= amount
    print("Withdrawal successful")
    print("Remaining balance:", balance)
else:
    print('error')    

```
