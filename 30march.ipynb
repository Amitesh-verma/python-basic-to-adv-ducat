```python
# menu - driven program (using while loop)
#  1 - add
#  2 - subtract
#  3 - Exit


num1 = int(input()) 
num2 = int(input())

operation = int(input('enter 1 for add,2 for sub,3 for exit'))

while operation <= 3:
    if operation == 1:
        print('sum of',num1, '+' ,num2, 'is', num1+num2)
        break
    elif operation == 2:
        print('sub of',num1, '-' ,num2, 'is', num1-num2)  
        break  
    elif operation == 3:
        print('exit')
        break
        
```
