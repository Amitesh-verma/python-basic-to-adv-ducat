```python
# def __init__ :- special method/magic method
# __init__ :- initilization/constructor
# rule :- constructor kehta hai ki jaise hi object banaoge to me automatically run 
class maggi:
    def __init__(self,masala,paani):
        self.masala=masala  #instance variable
        self.paani=paani
    def cook(self):
        print(f'maagi ban rahi hai {self.masala} ke sath {self.paani}')
    def google(self):
        print('hi google')
raman=maggi('hindustani masala','jal')
raman.google()
# jb ham kisi class jo variable ke ander store karte hai to wo object ban jata hai
# raman.cook()
# kabhi bhi agr aap method ko call karna chahte ho to object ke through karoge
```

```python
# Practice Questions – Attributes
# 1. Create a class FoodItem with class attribute category = "Snacks" and instance attribute name (“Samosa”, “GulabJamun”).

class FoodItem:
    category = "Snacks"
    def __init__(self,name):
        self.name = name
f1 = FoodItem('Samosa')

f2 = FoodItem('GulabJamun')   

print(f1.name,f1.category)
print(f2.name,f2.category)


```

```python
# . Create class Student that takes 3 marks and has a method average().

class Student:
    def __init__(self,name,listOfmarks):
        self.name = name
        self.listOfmarks = listOfmarks
    def Avg(self):
        sum = 0

        for eachValue in self.listOfmarks:
            sum = sum+eachValue

        Avg = sum/3
        print('avg in ',Avg)

student1 = Student('hello',([23,34,65])) 
    
student1.Avg()   


        
```

```python
# Static methods do not use self.

class Student:
    @staticmethod
    def school():
        print("ABC Public School") # generic name
Student.school()
```

```python
# Create static method to validate if a number is even.

class Num:
    @staticmethod
    def even(num):
        number=num % 2==0
        print('num is',number)
           
Num.even(2)        
        
```
