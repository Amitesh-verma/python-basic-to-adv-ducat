```python
rupee = 93.06
dollar = int(input())

conv = dollar * rupee

print(f'{dollar} after converting into INR : {conv:.2f}')
```

```python
# .format 

rupee = 93.06
dollar = int(input())

conv = dollar * rupee

print('{a} after converting into INR : {b}'.format(a = dollar,b=conv))
```
