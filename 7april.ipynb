# List Slicing Assignment


```python
# List [10, 20, 30, 40, 50, 60, 70]:
# • First 3 elements
# • Last 3 elements
# • Middle elements

list = [10, 20, 30, 40, 50, 60, 70]
list[0:3]   
list[-3:]
list[2:5]
```

```python
# List ko reverse karo using slicing
list = [10, 20, 30, 40, 50, 60, 70]
list[::-1]
```

```python
# Har second element nikaalo list me se
list = [10, 20, 30, 40, 50, 60, 70]
list[::2]
```

```python
# List ke first half aur second half alag-alag slice karo
lst = [10, 20, 30, 40, 50, 60, 70]
n=len(lst)
mid = n//2
lst[:mid]
lst[mid:]
```

```python
# Ek list me se:
# • Even index values
# • Odd index values
lst = [10, 20, 30, 40, 50, 60, 70]
print(lst[::2])
print(lst[1::2])    
```

```python
# List [1,2,3,4,5] me slicing ka use karke:
# • [2,3,4] banao
# • [5,4,3,2,1] banao

lst = [1,2,3,4,5]
print(lst[1:4])
print(lst[::-1])
```

```python
# Nested list me slicing apply karo
# Example: [[1,2],[3,4],[5,6]]

lst = [[1,2],[3,4],[5,6]]

print(lst[:2])
print(lst[-1:])
```
