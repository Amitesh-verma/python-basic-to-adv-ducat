```python
# 4. While Loop Questions
# Number reverse karo
# Digits count karo
# Sum of digits nikaalo
# Palindrome number check karo
# Fibonacci series print karo
```

```python
# Number reverse karo
n = int(input())

i = n
while i>=1:
    print(i)
    i = i-1      

```

```python
# Digits count karo

num =int(input())

count = 0
while num > 0:
    num = num // 10
    count =count+1
print(count)
```

```python
# Sum of digits nikaalo

num = int(input())

sum =0
while num >0:
    digit = num %10
    sum = sum + digit
    num = num//10   
print(sum)    
```

```python
# Palindrome number check karo
num = int(input())
org = num
rev = 0
while num > 0:
    digit  = num  % 10
    rev = rev * 10 + digit
    num = num//10
if org == rev:
    print('palindrome')
else:
    print('no palindrome')    
```

```python
# Fibonacci series print karo
num =int(input())
a =0
b =1
i =0
while i < num:
    print(a)
    next = a+b
    a =b
    b = next
    i= i+1
```

```python
# Number pattern (1, 12, 123…)
# 1
# 12
# 123
# 1234
# 12345

num = int(input())
for i in range(1,num+1):
    for j in range(1,i+1):
        print(j,end='')
    print() 
```

```python
# Inverted Pyramid pattern print karo
# *********
#  *******
#   *****
#    ***
#     *  

num = int(input())
for i in range(num, 0, -1):
    for j in range(num-i):
        print(' ',end ="")
    for j in range(2*i-1):
        print('*',end ="")
    print()
```

```python
# Pyramid pattern print karo
#     *
#    ***
#   *****
#  *******
# *********


num = int(input())
for i in range(num+1):
    for j in range(num-i):
        print(' ',end ="")
    for j in range(2*i-1):
        print('*',end ="")
    print()
```

```python
# Create a function that returns the largest of three numbers.
def largest_num(a,b,c):
    if a>=b and a>=c:
        return a
    elif b>=a and b>=c:
        return b
    else:
        return c
a = int(input('enter a'))
b = int(input('enter b'))
c = int(input('enter c'))


result = largest_num(a,b,c)
print('Largest num =',result)
```

```python
# Write a function to calculate the factorial of a number.

def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact*=i
    return fact
      
result = factorial(5)
print(result)      
```

```python
# Create a function that counts the number of vowels in a string.

def count_vowels(s):
    count=0
    for char in s:
        if char in "aeiouAEIOU":
            count +=1
    return count 

result = count_vowels('Hello world')

print(result)

                
```

```python
# Create a function that returns the sum of all elements in a list.
# → Loop and keep adding OR use sum().


def sum_list(nums):
    total = 0
    for i in nums:
        total+=i
    return total
    
result = sum_list([1,2,3])
print(result)
```

```python
# Write a function that finds the maximum number in a list.
# → Track a variable → update when you find a bigger number.

def find_max(nums):
    max_val = nums[0]

    for num in nums:
        if num > max_val:
            max_val = num
    return max_val

result = find_max([55,48,78,99,100])
print(result)       


```

```python
# Write a function to calculate the simple interest
# (SI = (P × R × T) / 100)
# Simple Interest
# → Direct formula → (P * R * T) / 100

def simple_int():
    p = int(input('Enter the profit value :'))
    r = int(input('Enter the rate value :'))
    t = int(input('Enter the time value :'))
    si = (p * r *t) / 100

    return si

result = simple_int()

print(result)
```

```python
# Create a function that checks if a number is prime.
# Prime number
# → Check divisibility from 2 to n-1

def prime():
    n = int(input('Enter the number :'))

    if n<=1:
        print('prime nahi hai')
        return

    for i in range(2,n):
          if n % i == 0:
            print('prime nahi hai ')
            return 
          
    print('prime hai')        

prime()
```
