```python
# syntax
# try: block
# except block
# else block
# finally block

# rules
# try block  error fetch karne ka kaam

# except block :- agr error fetch hota hai try ke ander then except block custom error show karega

# else block :- ye kehta agr try block ne error fetch nhi kiya then try ke sath only else chalta hai

# finally:- ye kehta block error fetch kare ya nhi me to hamesha run karta hoon
```

```python
# practical things
try:
    print('chlo maggi banate hai')
    paani=int(input('paani rakho')) #type cast string to int
    maggi=int(input('kitne packet bhai'))
    cooking=maggi/paani
    print('ye maggie to me hi khaunga')
except:
    print('paani nhi hai bhai paani to rakho pagal')
else: #else depent on try true
    print('chlo pet bhar gya thanku')
finally:  #independent hai 
    print('program khatm ghar jao')
```
